```python
import matplotlib.pyplot as plt

marks = [40, 30, 20, 10]

plt.pie(marks)

plt.show()
```

```python
subjects = ["Math", "Science", "English", "Computer"]
marks = [40, 30, 20, 10]

plt.pie(marks, labels=subjects)

plt.show()
```

```python
subjects = ["Math", "Science", "English", "Computer"]
marks = [40, 30, 20, 10]

plt.pie(
    marks,
    labels=subjects,
    colors=["yellow", "blue", "green", "orange"]
)

plt.show()
```

```python
subjects = ["Math", "Science", "English", "Computer"]
marks = [40, 30, 20, 10]

plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%"
)

plt.show()
```

```python
fruits = ["Apple", "Banana", "Orange", "Grapes"]
sales = [35, 25, 20, 20]

plt.pie(sales, labels=fruits,
        colors=["red", "yellow", "orange", "purple"],
        autopct="%1.1f%%")
plt.title("Fruit Sales")
plt.show()
```

```python
fruits = ["Apple", "Banana", "Orange", "Grapes"]
sales = [35, 25, 20, 20]

plt.pie(
    sales,
    labels=fruits,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Fruit Sales")
plt.show()
```

```python
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [40, 30, 20, 10]

plt.pie(sales, labels=products,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Product sales")
plt.show()
```

```python
explode = [0.1, 0.3, 0.2, 0]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    explode=explode
)
```

```python
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [40, 30, 20, 10]

explode = [0.1, 0, 0, 0]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    explode=explode
)

plt.title("Product Sales")
plt.show()
```

```python
subjects = ["Math", "Physics", "Chemistry", "CS"]
marks = [30, 25, 20, 25]
explode = [0.1, 0, 0, 0]

plt.pie(marks, labels=subjects,
        explode=explode, 
        autopct="%1.1f%%",
        startangle=90,
        shadow=True)
plt.title("Subject Distribution")
plt.show()
```

```python
plt.pie(
    marks,
    labels=subjects,
    explode=explode,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True
)
```

```python
plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"edgecolor": "black", "linewidth": 2}
)
```

```python
subjects = ["Math", "Physics", "Chemistry", "CS"]
marks = [30, 25, 20, 25]
explode = [0.1, 0, 0, 0]

plt.pie(marks, labels=subjects, 
        autopct="%1.1f%%",
        startangle=90,
        explode=explode,
        shadow=True,
        wedgeprops={"edgecolor" : "black", "linewidth" : 1})
plt.title("Subject Distribution")
plt.show()
```

```python
plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%",
    pctdistance=0.7
)
```

```python
subjects = ["Math", "Physics", "Chemistry", "CS"]
marks = [30, 25, 20, 25]
explode = [0.1, 0, 0, 0]

plt.pie(marks, labels=subjects, 
        autopct="%1.1f%%",
        startangle=90,
        explode=explode,
        shadow=True,
        pctdistance=0.8,
        wedgeprops={"edgecolor" : "black", "linewidth" : 1})
plt.title("Subject Distribution")
plt.show()
```

```python
expenses = ["Food", "Rent", "Travel", "Shopping", "Other"]
amount = [2500, 8000, 2000, 1500, 1000]
explode=[0, 0.05, 0, 0, 0]

plt.pie(amount, labels=expenses,
        autopct="%1.1f%%",
        startangle=90,
        explode=explode,
        shadow=True,
        wedgeprops={"edgecolor":"black", "linewidth":2},
        pctdistance=0.7)
plt.title("Monthly Expenses")
plt.show()
```
