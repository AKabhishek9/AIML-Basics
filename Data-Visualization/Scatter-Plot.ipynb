```python
import matplotlib.pyplot as plt

hours = [1,2,3,4,5,6,7,8]
marks = [35,40,50,55,65,72,80,90]

plt.scatter(hours, marks, color="red", s=100, marker="*")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
temperature = [20,22,24,26,28,30,32]
iceCreamSales = [50,60,68,80,95,110,125]

plt.scatter(temperature, iceCreamSales,
            color="blue",
            s=120,
            marker="^",
            edgecolors="black")
plt.title("Temperature vs Ice Cream Sales")
plt.xlabel("Temperature")
plt.ylabel("Sales")
plt.show()
```

```python
age = [18,20,22,24,26,28,30]
salary = [20,24,28,35,40,48,55]

plt.scatter(age, salary,
            color="green",
            s=100,
            marker="s",
            edgecolors="black")
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
import matplotlib.pyplot as plt

boys = [150,155,160,165,170]
boysWeight = [45,50,55,60,65]

girls = [145,150,155,160,165]
girlsWeight = [42,46,50,54,58]

plt.scatter(
    boys,
    boysWeight,
    color="blue",
    label="Boys",
    s=100
)

plt.scatter(
    girls,
    girlsWeight,
    color="red",
    label="Girls",
    s=100
)

plt.title("Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.legend()
plt.show()
```

```python
teamA_x = [10,20,30,40,50]
teamA_y = [15,25,35,45,55]

teamB_x = [12,22,32,42,52]
teamB_y = [18,28,38,48,58]

plt.scatter(teamA_x, teamA_y,
            color="blue",
            marker="o",
            label="Team A",
            s=100)
plt.scatter(teamB_x, teamB_y,
            color="red",
            marker="^",
            label="Team B",
            s=100)

plt.title("Team A vs Team B")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,15,30,25]

colors = [10,20,30,40,50]

plt.scatter(
    x,
    y,
    c=colors,
    cmap="plasma",
    s=150
)

plt.colorbar()

plt.show()
```

```python
hours = [1,2,3,4,5,6]
marks = [40,50,60,65,75,90]
performance = [1,2,3,4,5,6]

plt.scatter(hours, marks,
            c=performance,
            cmap="plasma",
            s=120)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.colorbar()
plt.show()
```
